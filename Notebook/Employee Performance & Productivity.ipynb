```python
# Importing essential data manipulation and visualization libraries for the analysis.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the employee performance dataset and taking a quick look at the first few rows.
data=pd.read_csv("C:\\Users\\Dell\\OneDrive\\Desktop\\Project 1\\Extended_Employee_Performance_and_Productivity_Data\\Extended_Employee_Performance_and_Productivity_Data.csv", encoding="latin1")
data
```

```python
# Cleaning up column names by removing underscores for better readability and querying.
data.columns=data.columns.str.replace("_", " ")
data.columns
```

```python
# Counting total employee distribution across individual company departments.
data["Department"].value_counts()
```

```python
# Calculating high-level workforce KPIs like overall headcount, attrition rate, and averages.
total_employee=data["Employee ID"].count()
attrition_rate=data["Resigned"].sum()/total_employee
avg_monthly_salary=data["Monthly Salary"].mean()
avg_satisfaction_score=data["Employee Satisfaction Score"].mean()
avg_performance_score=data["Performance Score"].mean()
avg_overtime_hours=data["Overtime Hours"].mean()
avg_tenure_yrs=data["Years At Company"].mean()
total_department=data["Department"].nunique()
```

```python
# Displaying high-level workforce KPIs like overall headcount, attrition rate, and averages.
print("="*40)
print("""----------Workforce Overview &----------
-------Key Performance Indicators-------""")
print("="*40)
print()
print(f"Total Employees: {total_employee}")
print("Number of Departments: ",total_department)
print(f"Average Tenure Years: {avg_tenure_yrs:,.2f}Y")
print(f"Attrition Rate: {attrition_rate*100:,.2f}%")
print(f"Average Monthly Salary: ${avg_monthly_salary:,.2f}")
print(f"Average Satisfaction Score: {round(avg_satisfaction_score):,.2f}")
print(f"Average Performance Score: {round(avg_performance_score):,.2f}")
print(f"Average Overtime Hours: {avg_overtime_hours:,.2f} Hours")
```

```python
# Visualizing the employee distribution across different departments to check headcount balance.
plt.figure(figsize=(14,6))
category_dep=data.groupby("Department")["Employee ID"].count().sort_values(ascending=False).reset_index()

ax=sns.barplot(
    x="Department",
    y="Employee ID",
    data=category_dep,
    legend=False,
    color="c"

)

for i in ax.containers:
    ax.bar_label(
        i,
        fmt="%.0f",
        padding=1,
        fontweight="bold",
        fontsize=9
)

plt.title(
    "Total Employee by Department",
    fontsize=14,
    fontweight="bold"
)
plt.ylabel(
    "Count of Employee"
)
plt.xlabel(
    "Department"
)
plt.savefig("01total_employee_by_department.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Analyzing department-wise attrition rates to identify where employee turnover is highest.
plt.figure(figsize=(14,6))
attrition_rate=(data.groupby("Department")["Resigned"].sum()/
                data.groupby("Department")["Resigned"].count()*100).sort_values(ascending=False).reset_index(name="Attrition_Rate")


ax=sns.barplot(
    x="Department",
    y="Attrition_Rate",
    data=attrition_rate,
    legend=False,
    color="cyan"
)


for i in ax.containers:
    ax.bar_label(
        i,
        fmt="%.2f%%",
        padding=3,
        fontweight="bold",
        fontsize=9
    )

plt.title(
    "Attrition Rate by Department",
    fontweight="bold",
    fontsize=14
)
plt.xlabel(
    "Department"
)
plt.ylabel(
    "Attrition Rate"
)
plt.savefig("02_attrition_rate_by_department.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Checking average monthly salaries department-wise to ensure internal pay equity.
plt.figure(figsize=(14,6))
avg_dep_monthly_salary=data.groupby("Department")["Monthly Salary"].mean().sort_values(ascending=False).reset_index(name="Salary")

ax=sns.barplot(
    x="Department",
    y="Salary",
    data=avg_dep_monthly_salary,
    legend=False,
    color="orchid"
)

for i in ax.containers:
    ax.bar_label(
        i,
        fmt="$%.2f",
        padding=0,
        fontweight="bold",
        fontsize=9
    )

plt.title(
    "AVG Monthly Salary by Department",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel(
    "Department"
)
plt.ylabel(
    "Salary (USD)"
)


plt.savefig("03_avg_monthly_salary_by_department.png", dpi=300, bbox_inches="tight")        
plt.show()
```

```python
# Plotting employee satisfaction score distribution to gauge overall workforce sentiment.
plt.figure(figsize=(6,4))

# Creating satisfaction score bins/ranges to categorize employee sentiment levels.
bins=[0,1,2,3,4,5]
labels=[1,2,3,4,5]
data["Satisfaction Range"]=pd.cut(
    data["Employee Satisfaction Score"],
    bins=bins,
    labels=labels,
    include_lowest=True,
)

employee_satisfaction_score=data.groupby("Satisfaction Range", observed=False)["Employee ID"].count().sort_values(ascending=False).reset_index(name="Employees")

ax=sns.barplot(
    x="Satisfaction Range",
    y="Employees",
    data=employee_satisfaction_score,
    legend=False,
    color="gold"
)

for i in ax.containers:
    ax.bar_label(
        i,
        fmt="%.0f",
        padding=0,
        fontweight="bold",
        fontsize=9
    )

plt.title(
    "Employee Satisfaction Score",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel(
    "Satisfaction Score"
)
plt.ylabel(
    "Employee"
)


plt.savefig("08_employee_satisfaction_score.png", dpi=300, bbox_inches="tight")        
plt.show()
```

```python
# Mapping and comparing satisfaction scores between active employees and those who resigned.
def resign(x):
    if x:
        return "Resigned"
    else:
        return "Stayed"
data["Resignation"]=data["Resigned"].apply(resign)

avg=data.groupby("Resignation")["Employee Satisfaction Score"].mean().sort_values(ascending=False).reset_index(name="Average Score")
avg
```

```python
# Comparing satisfaction scores between employees who stayed versus those who resigned.
plt.figure(figsize=(6,4))

ax=sns.barplot(
    x="Resignation",
    y="Average Score",
    data=avg,
    legend=False,
    color="lightpink"
);

for i in ax.containers:
    ax.bar_label(
        i,
        fmt="%.2f",
        padding=0,
        fontweight="bold",
        fontsize=9
    )

plt.title(
    "AVG Satisfaction: Stayed vs Resigned",
    fontweight="bold",
    fontsize=14
)

plt.ylabel(
    "AVG Satisfaction Score"
)

plt.savefig("09_avg_satisfaction_resignation.png", dpi=300, bbox_inches="tight")
plt.show()
```
